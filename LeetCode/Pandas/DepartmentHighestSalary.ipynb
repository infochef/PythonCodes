```python
import pandas as pd

data = [[1, 'Joe', 70000, 1], [2, 'Jim', 90000, 1], [3, 'Henry', 80000, 2], [4, 'Sam', 60000, 2], [5, 'Max', 90000, 1]]
employee = pd.DataFrame(data, columns=['id', 'name', 'salary', 'departmentId']).astype({'id':'Int64', 'name':'object', 'salary':'Int64', 'departmentId':'Int64'})
data = [[1, 'IT'], [2, 'Sales']]
department = pd.DataFrame(data, columns=['id', 'name']).astype({'id':'Int64', 'name':'object'})
```

```python
df = pd.merge(employee, department, left_on='departmentId', right_on='id', how='left', suffixes=('_emp','_dept'))
df2 = employee.groupby('departmentId')['salary'].max().reset_index().rename(columns={'salary':'max_salary'})
df3 = df.merge(df2, on='departmentId', how = 'left')
df4 = df3[df3['salary']==df3['max_salary']][['name_dept','name_emp','salary']].rename(columns={'name_dept':'Departement', 'name_emp':'Employee', 'salary':'Salary'})
print(df4)
```
